# PySR sanity-check notebook: recover a known function

Goal: generate synthetic data from a known function

\[
y = \sqrt{x} + 2x
\]

add Gaussian noise, then **hide the true function** and let **PySR** rediscover an equation from data.

You get:
- A plot of the noisy data, the true function, and PySR's best-fit expression
- A slider-driven UI (via `ipywidgets`) with a **Run PySR** button so you can test different noise levels and sample sizes


```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pysr import PySRRegressor

import ipywidgets as widgets
from IPython.display import display, clear_output
```

```
def true_function(x: np.ndarray) -> np.ndarray:
    # Ground-truth function (used only to generate synthetic data)
    return np.sqrt(x) + 2.0 * x
```

```
def make_synthetic_data(n_points: int, noise_sigma: float, x_min: float = 0.0, x_max: float = 10.0, seed: int = 0):
    rng = np.random.default_rng(seed)
    x = np.linspace(x_min, x_max, n_points)
    y_clean = true_function(x)
    y_noisy = y_clean + rng.normal(loc=0.0, scale=noise_sigma, size=n_points)
    return x, y_clean, y_noisy
```

```
def fit_pysr(x: np.ndarray, y: np.ndarray, niterations: int, maxsize: int, seed: int = 0):
    X = x.reshape(-1, 1)

    model = PySRRegressor(
        niterations=niterations,
        maxsize=maxsize,
        binary_operators=["+", "-", "*", "/"],
        unary_operators=["sqrt"],
        model_selection="best",
        random_state=seed,
        verbosity=0,
    )
    model.fit(X, y)
    return model
```

```
def plot_results(x, y_clean, y_noisy, model, x_grid):
    y_pred = model.predict(x_grid.reshape(-1, 1))

    plt.figure(figsize=(8, 5))
    plt.scatter(x, y_noisy, s=18, label="Noisy data")
    plt.plot(x, y_clean, linewidth=2, label="True function")
    plt.plot(x_grid, y_pred, linewidth=2, label="PySR prediction")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("PySR recovering y = sqrt(x) + 2x from noisy samples")
    plt.legend()
    plt.show()

    residuals = y_noisy - model.predict(x.reshape(-1, 1))
    rmse = float(np.sqrt(np.mean(residuals**2)))
    return rmse
```

```
# Interactive controls
n_points_slider = widgets.IntSlider(value=120, min=20, max=400, step=10, description="n_points")
noise_slider = widgets.FloatSlider(value=0.2, min=0.0, max=2.0, step=0.05, description="noise σ")
niter_slider = widgets.IntSlider(value=200, min=50, max=800, step=50, description="iterations")
maxsize_slider = widgets.IntSlider(value=18, min=8, max=40, step=2, description="maxsize")
seed_slider = widgets.IntSlider(value=0, min=0, max=50, step=1, description="seed")

run_button = widgets.Button(description="Run PySR", button_style="primary")
out = widgets.Output()

ui = widgets.VBox([
    widgets.HBox([n_points_slider, noise_slider]),
    widgets.HBox([niter_slider, maxsize_slider, seed_slider]),
    run_button,
    out
])

display(ui)

def on_run_clicked(_):
    with out:
        clear_output(wait=True)

        n_points = int(n_points_slider.value)
        noise_sigma = float(noise_slider.value)
        niterations = int(niter_slider.value)
        maxsize = int(maxsize_slider.value)
        seed = int(seed_slider.value)

        x, y_clean, y_noisy = make_synthetic_data(n_points=n_points, noise_sigma=noise_sigma, seed=seed)

        # "Take away the function": from here on, we only give PySR (x, y_noisy).
        model = fit_pysr(x=x, y=y_noisy, niterations=niterations, maxsize=maxsize, seed=seed)

        x_grid = np.linspace(x.min(), x.max(), 500)
        rmse = plot_results(x, y_clean, y_noisy, model, x_grid)

        print("Selected equation (best tradeoff):")
        print(model)
        print()
        print(f"Train RMSE against noisy samples: {rmse:.6g}")
        print()
        print("Top 10 discovered equations (score-sorted):")
        display(model.equations_.sort_values('score', ascending=False).head(10)[['score','loss','complexity','equation']])

run_button.on_click(on_run_clicked)
```

## Tips
- Increase **iterations** if PySR doesn’t recover the structure at higher noise.
- Decrease **maxsize** if equations get too complex/ugly.
- Try a few **seeds** to see stability.
